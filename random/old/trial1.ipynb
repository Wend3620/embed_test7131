```python
import xarray as xr, os, numpy as np, matplotlib.pyplot as plt
from datetime import datetime, timedelta
import json
```

```python
start = datetime.strptime("02_26", "%m_%y")
end = datetime.strptime("02_26", "%m_%y")

months = []
current = start

while current <= end:
    months.append(current.strftime("%m_%y"))
    if current.month == 12:
        current = current.replace(year=current.year + 1, month=1)
    else:
        current = current.replace(month=current.month + 1)

bins = [months[i:i + 3] for i in range(0, len(months), 3)]
seasons = ['JJA', 'SON', 'DJF', 'MAM', 'JJA', 'SON', 'DJF', 'MAM']
bins
```

```python
folders = []
data_folder = "/data/wdu36"
for sat in ['sat1', 'sat2']:
    for months, key in zip(bins, seasons):
        date = {}
        export_name = f'{sat}_{key}_{months[0][-2:]}'
        for month in months:
            folder_name = f'{sat}_{month}'
            try: 
                os.listdir(f'{data_folder}/{folder_name}')
            except FileNotFoundError:
                print(f"Folder {folder_name} not found. Skipping.")
                folders.append(folder_name)
                continue
                
            for file in os.listdir(f'{data_folder}/{folder_name}'):
                if file.endswith('.nc'):
                    time = datetime.strptime(file.split('_')[5], '%Y%m%d%H%M%S')
                    
                    time_str = time.strftime('%Y-%m-%d')
                    if time_str not in date:
                        date[time_str] = 1
                    else:
                        date[time_str] += 1
                
        with open(f'{data_folder}/{export_name}.json', 'w', encoding='utf-8') as f:
            json.dump(date, f, ensure_ascii=False, indent=2)
        with open(f'./data/{export_name}.json', 'w', encoding='utf-8') as f:
            json.dump(date, f, ensure_ascii=False, indent=2)
        print(f"Saved to date.json for folder {export_name}.json")
print(f'Folders missing: {folders}')
```
